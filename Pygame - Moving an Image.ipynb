```python
import pygame
import sys

# 1. Initialize Pygame
pygame.init()

# 2. Setup the Display
screen_width = 800
screen_height = 600
screen = pygame.display.set_mode((screen_width, screen_height))
pygame.display.set_caption("Smooth Movement Example")

# 3. Load Image (or create a fallback surface)
try:
    player_image = pygame.image.load("character.png").convert_alpha()
except:
    # If no image found, create a 50x50 blue square
    player_image = pygame.Surface((50, 50))
    player_image.fill((0, 120, 255))

# 4. Movement Variables
player_x = 375
player_y = 275
player_speed = 5

# 5. Game Clock (to control the frame rate)
clock = pygame.time.Clock()

# --- Main Game Loop ---
running = True
while running:
    # A. Event Handling (Input)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # B. Movement Logic (Polling)
    # Using get_pressed() allows for smooth, continuous movement
    keys = pygame.key.get_pressed()
    
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= player_speed
    if keys[pygame.K_RIGHT] and player_x < screen_width - player_image.get_width():
        player_x += player_speed
    if keys[pygame.K_UP] and player_y > 0:
        player_y -= player_speed
    if keys[pygame.K_DOWN] and player_y < screen_height - player_image.get_height():
        player_y += player_speed

    # C. Rendering (Drawing)
    screen.fill((30, 30, 30))  # Clear screen with dark gray
    screen.blit(player_image, (player_x, player_y))  # Draw the image
    
    pygame.display.flip()  # Update the display

    # D. Maintain 60 Frames Per Second
    clock.tick(60)

pygame.quit()
sys.exit()
```
